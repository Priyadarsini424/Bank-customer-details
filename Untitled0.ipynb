```python
import pandas as pd
data= pd.read_csv('Churn_Modelling.csv')
#display top 5 rows of the dataset
data.head()
#display last 5 rows of the dataset
data.tail()
#find shape of our dataset(Number of rows and number of columns)
data.shape
print("Number of rows",data.shape[0])
print("Number of Columns",data.shape[1])
#Get information about our dataset like total number of rows , total number of columns, datatypes of each column and memory requirement
data.info()


```

```python
#Check null alues in the dataset
data.isnull()
data.isnull().sum()
#Get overall statistics about the dataset
data.describe()
data.describe(include='all')
```

```python
#dropping irrelevant features
data.columns
data= data.drop(['RowNumber','CustomerId','Surname'],axis=1)
data.head()
```

```python
#Encoding categorial data
data['Geography'].unique()
data= pd.get_dummies(data,drop_first=True)
data.head()

```

### Note on `NameError`

If you encounter a `NameError` like `name 'X' is not defined`, please ensure that you have run the preceding cell (`qMOabCgXfoN6`) which defines the variables `X` and `Y`.

```python
#not handling imbalanced
data['Exited'].value_counts()
import seaborn as sns
sns.countplot(x='Exited', data=data)
X= data.drop('Exited',axis =1)
Y= data['Exited']
```

```python
import pandas as pd
import numpy as np
#splitting the dataset into the training set and test set
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test= train_test_split(X,Y, test_size=0.20, random_state=42, stratify=Y)
Y_train
```

```python
#feature scaling
from sklearn.preprocessing import StandardScaler
sc= StandardScaler()
X_train= sc.fit_transform(X_train)
X_test= sc.transform(X_test)
Y_train
```

```python
#logostic regreeion
from sklearn.linear_model import LogisticRegression
classifier= LogisticRegression()
classifier.fit(X_train,Y_train)
Y_pred= classifier.predict(X_test)
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print(accuracy_score(Y_test,Y_pred))
print(confusion_matrix(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
```

```python
#handling imbalanced data with SMOTE
from imblearn.over_sampling import SMOTE
sm= SMOTE()
X_train_sm, Y_train_sm= sm.fit_resample(X_train,Y_train)
#print(X_train_sm.shape)
#value_counts= Y_train_sm.value_counts()
#print(value_counts)
#value_counts= X_train_sm.value_counts()
#print(value_counts)
Y_train_sm.value_counts()

```

```python
#SUPPORT VECTOR CLASSIFIER
from sklearn.svm import SVC
classifier= SVC()
classifier.fit(X_train_sm,Y_train_sm)
Y_pred= classifier.predict(X_test)
print(accuracy_score(Y_test,Y_pred))
print(confusion_matrix(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))

```

```python
#KNeighbours classifier
from sklearn.neighbors import KNeighborsClassifier
classifier= KNeighborsClassifier()
classifier.fit(X_train_sm,Y_train_sm)
Y_pred= classifier.predict(X_test)
print(accuracy_score(Y_test,Y_pred))
print(confusion_matrix(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
```

```python
#Decision tree classifier
from sklearn.tree import DecisionTreeClassifier
classifier= DecisionTreeClassifier()
classifier.fit(X_train_sm,Y_train_sm)
Y_pred= classifier.predict(X_test)
print(accuracy_score(Y_test,Y_pred))
print(confusion_matrix(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
```

```python
#Random Forest classifier
from sklearn.ensemble import RandomForestClassifier
classifier= RandomForestClassifier()
classifier.fit(X_train_sm,Y_train_sm)
Y_pred= classifier.predict(X_test)
print(accuracy_score(Y_test,Y_pred))
print(confusion_matrix(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
```

```python
#Gradient Boosting Classifier
from sklearn.ensemble import GradientBoostingClassifier
classifier= GradientBoostingClassifier()
classifier.fit(X_train_sm,Y_train_sm)
Y_pred= classifier.predict(X_test)
print(accuracy_score(Y_test,Y_pred))
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
from sklearn.preprocessing import StandardScaler
import joblib
import pandas as pd
import numpy as np

# --- Create Input Widgets ---
# The features are: CreditScore, Age, Tenure, Balance, NumOfProducts, HasCrCard, IsActiveMember, EstimatedSalary,
# Geography_Germany, Geography_Spain, Gender_Male.

# Define widgets for each input feature
w_credit_score = widgets.IntText(value=600, description='CreditScore:', min=350, max=850)
w_age = widgets.IntText(value=35, description='Age:', min=18, max=92)
w_tenure = widgets.IntText(value=5, description='Tenure:', min=0, max=10)
w_balance = widgets.FloatText(value=75000.0, description='Balance:')
w_num_products = widgets.IntText(value=1, description='NumOfProducts:', min=1, max=4)
w_has_cr_card = widgets.Dropdown(options={('No', 0), ('Yes', 1)}, value=1, description='HasCrCard:')
w_is_active_member = widgets.Dropdown(options={('No', 0), ('Yes', 1)}, value=1, description='IsActiveMember:')
w_estimated_salary = widgets.FloatText(value=100000.0, description='EstimatedSalary:')
w_geography = widgets.Dropdown(
    options={'France': 3, 'Germany': 1, 'Spain': 2}, # Map text to integer as used in original logic
    value=3,
    description='Geography:'
)
w_gender = widgets.Dropdown(options={'Female': 0, 'Male': 1}, value=0, description='Gender:')

# Create a button for prediction
predict_button = widgets.Button(description="Predict Churn")

# Create an output widget to display results
output_area = widgets.Output()

# --- Prediction Function ---
def on_predict_button_clicked(b):
    with output_area:
        clear_output(wait=True)
        try:
            # Retrieve values from widgets
            credit_score = w_credit_score.value
            age = w_age.value
            tenure = w_tenure.value
            balance = w_balance.value
            num_of_products = w_num_products.value
            has_cr_card = w_has_cr_card.value
            is_active_member = w_is_active_member.value
            estimated_salary = w_estimated_salary.value

            # Handle Geography (1=Germany, 2=Spain, 3=France)
            geography_input = w_geography.value
            geo_germany = 0
            geo_spain = 0
            if geography_input == 1: # Germany
                geo_germany = 1
            elif geography_input == 2: # Spain
                geo_spain = 1
            # For France (geography_input == 3), both remain 0

            # Handle Gender (0=Female, 1=Male)
            gender_male = w_gender.value

            # Create a DataFrame for the new customer data with correct column order
            # X.columns and sc are assumed to be defined and fitted in previous cells
            # from the notebook state. If not, this cell needs to be run after X and sc are defined.
            new_customer_data_df = pd.DataFrame([[
                credit_score, age, tenure, balance, num_of_products, has_cr_card,
                is_active_member, estimated_salary, geo_germany, geo_spain, gender_male
            ]], columns=X.columns)

            # Scale the new data using the already fitted StandardScaler 'sc'
            scaled_new_customer = sc.transform(new_customer_data_df.values)

            # Load the model (assuming 'Churn_Modelling.pkl' is saved and accessible)
            model = joblib.load('Churn_Modelling.pkl')

            # Make prediction
            result = model.predict(scaled_new_customer)

            if result[0] == 0:
                print("Prediction: No Churn")
            else:
                print("Prediction: Churn")

        except NameError as ne:
            print(f"Error: {ne}. Make sure 'X' and 'sc' are defined in previous cells.")
        except FileNotFoundError:
            print("Error: 'Churn_Modelling.pkl' not found. Ensure the model is saved correctly.")
        except ValueError as ve:
            print(f"Input Error: {ve}. Please ensure all fields have valid numerical inputs.")
        except Exception as e:
            print(f"An unexpected error occurred: {e}")

# Attach the function to the button click event
predict_button.on_click(on_predict_button_clicked)

# --- Display Widgets ---
display(
    w_credit_score,
    w_age,
    w_tenure,
    w_balance,
    w_num_products,
    w_has_cr_card,
    w_is_active_member,
    w_estimated_salary,
    w_geography,
    w_gender,
    predict_button,
    output_area
)
```
